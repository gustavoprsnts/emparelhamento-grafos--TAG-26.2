### Importação dos dados para o projeto

```python
import re
# Abre o arquivo de dados 
with open("../database/dados.txt", "r", encoding="utf-8") as arquivo:
    linhas = arquivo.readlines()

# Organizando os projetos
projetos = {}
for linha in linhas:
    linha = linha.strip()
    # (código projeto, número de vagas, requisito mínimo de notas para vagas)
    padrao = re.fullmatch(r"\(P(\d+),\s*(\d+),\s*(\d+)\)",linha)
    if padrao:
        numero, vagas, nota_minima = map(int, padrao.groups())
        projetos[f"P{numero}"] = {
            "vagas": vagas,
            "nota_minima": nota_minima
        }
# Verificando se deu certo
print(f"Total de projetos: {len(projetos)}")
print(projetos)
```

```python
# Organizando os alunos
alunos = {}
for linha in linhas:
    linha = linha.strip()
    # (código aluno):(projetos preferenciais na ordem) (Nota do aluno)
    padrao = re.fullmatch(r"\(A(\d+)\):\s*\(([^)]*)\)\s*\((\d+)\)",linha)
    if padrao:
        numero, preferencias, nota = padrao.groups()
        lista_preferencias = [projeto.strip() for projeto in preferencias.split(",")]
        alunos[f"A{numero}"] = {
            "preferencias": lista_preferencias,
            "nota": int(nota)
        }
# Testando para vê se deu certo
print(f"Total de alunos: {len(alunos)}")
print(alunos)
```

```python
# Limpando os dados inconsistentes do txt
alunos_limpos = {}
for aluno, dados in alunos.items():
    preferencias_validas = []
    for projeto in dados["preferencias"]:
        # Ignora projetos inexistentes - P51 ...
        if projeto not in projetos:
            continue
        # Ignora preferências repetidas
        if projeto in preferencias_validas:
            continue
        preferencias_validas.append(projeto)

    alunos_limpos[aluno] = {
        "preferencias": preferencias_validas,
        "nota": dados["nota"]
    }
# Testando para vê se deu certo
print(f"Total de alunos: {len(alunos_limpos)}")
print(alunos_limpos)
```

### Implementação do algoritmo de gale-shapley
Temos 2 grupos: um de projetos acadêmicos e outra de alunos interessados nesses projetos. Cada aluno tem uma lista de preferência de projetos (do que mais quer ao que menos quer). Os projetos possuem uma nota mínima para aceitar o aluno. O objetivo do algoritmo é formar pares de modo que o resultado seja estável.

-> Passo a passo do algoritmo:

1. Todos começam livres. Cada aluno propõem para o projeto mais bem colocado em sua lista que ele ainda não tentou.
2. Se o projeto tiver vaga e o aluno tiver uma nota >= nota_minima -> aceita
3. Se o aluno tiver uma nota < nota_minima -> rejeita
4. Se o projeto não tem mais vagas e nota >= nota_minima, então compara os alunos que estão ocupando essas vagas com quem está propondo<br>
    * 4.1 Se o projeto preferir os estudantes atuais -> rejeita<br>
    * 4.2 Se o projeto preferir o novo, então aceita e tira o antigo
5. Repete até ninguém ficar livre ou até encontrar o emparelhamento máximo

```python
from collections import deque

# Função para a comparação entre alunos
def prioridade(aluno, alunos):
    # Pega a nota do aluno solicitado
    nota = alunos[aluno]["nota"]
    # Pega o numero do aluno -> criterio de desempate
    numero = int(aluno[1:])
    return (nota,-numero)

def gale_shapley(projetos,alunos):
    # Todos os alunos iniciam livres
    fila = deque(alunos.keys())
    # Guarda a próxima preferencia que será tentada pelo aluno
    prox_pref = {
        aluno: 0 for aluno in alunos
    }
    # O projeto que o aluno pegou
    alocado = {
        aluno: None for aluno in alunos
    }
    # Os alunos que estão preenchendo as vagas do projeto
    selecionados = {
        projeto: [] for projeto in projetos
    }
    # Histórico para a visualização posterior
    historico = []

    while fila:
        aluno = fila.popleft() # Pega o 1° aluno da fila
        preferencias = alunos[aluno]["preferencias"]
        # Verifica se tem mais projeto para olhar
        if prox_pref[aluno] >= len(preferencias):
            continue
        # Passo 1: O aluno propoem ao 1° projeto mais bem colocado que não tentou
        projeto = preferencias[prox_pref[aluno]]
        prox_pref[aluno] += 1
        historico.append((aluno, projeto, "proposta"))

        nota = alunos[aluno]["nota"]
        nota_minima = projetos[projeto]["nota_minima"]
        vagas = projetos[projeto]["vagas"]

        # Passo 3: se o aluno não tem nota suficiente rejeita
        if nota < nota_minima:
            historico.append((aluno, projeto, "rejeitado_nota"))
        # Passo 2: se o projeto tem vaga, só aceita
        elif len(selecionados[projeto]) < vagas:
            selecionados[projeto].append(aluno)
            alocado[aluno] = projeto
            historico.append((aluno, projeto, "aceito"))
        # Passo 4: não tem vaga -> comparação
        else:
            # Encontra o pior aluno atualmente aceito
            pior_aluno = min(
                selecionados[projeto],
                key=lambda a: prioridade(a, alunos)
            )
            # Passo 4.2: novo aluno é melhor
            if prioridade(aluno,alunos) > prioridade(pior_aluno,alunos):
                selecionados[projeto].remove(pior_aluno)
                selecionados[projeto].append(aluno)

                alocado[pior_aluno] = None
                alocado[aluno] = projeto
                historico.append((aluno, projeto, "substituiu", pior_aluno))
                #Quem foi removido volta para a fila
                fila.append(pior_aluno)

            # Passo 4.1: os alunos atuais são melhores
            else:
                historico.append((aluno, projeto, "rejeitado_preferencia"))
        # Se o aluno não conseguiu nenhum projeto, ele volta para a fila
        if alocado[aluno] is None:
            if prox_pref[aluno] < len(preferencias):
                fila.append(aluno)
                
    return alocado, selecionados, historico
```

### Gale-Shapley para os projetos
1. Escolhe um aluno legivel da sua lista que ainda não fez proposta
2. Se o aluno ta livre -> aceita
3. Se o aluno ta com projeto, faz a comparação entre ambos os projetos pela lista de preferencia do aluno<br>
    * 3.1. Se o projeto novo é o preferido, aceita e libera o outro 
    * 3.2. Se o atual é o preferido, rejeita


```python
# Preferência na comparação entre projetos
def prioridade_proj(alunos,projetos):
    preferencias = {}
    for projeto,dados in projetos.items():
        candidatos=[]
        for aluno,info in alunos.items():
            # O aluno deve estar interessado no projeto
            if projeto not in info["preferencias"]:
                continue
            # Verifica a nota mínima
            if info["nota"] < dados["nota_minima"]:
                continue
            candidatos.append(aluno)
        # Ordena os alunos de acordo com as preferencias dos projetos
        candidatos.sort(
        key=lambda a: prioridade(a, alunos),
        reverse=True
        )
        preferencias[projeto] = candidatos
    return preferencias

def gale_shapley_proj(alunos,projetos):
    # Preferencias de cada projeto
    preferencias = prioridade_proj(alunos, projetos)
    # Projetos começam na fila de propostas
    fila = deque(projetos.keys())
    # Próximo aluno que cada projeto tentará
    proximo_candidato = {
        projeto: 0 for projeto in projetos
    }
    # Projeto em que cada aluno está alocado
    alocado = {
        aluno: None for aluno in alunos
    }
    # Alunos selecionados em cada projeto
    selecionados = {
        projeto: [] for projeto in projetos
    }
    historico = []
    while fila:
        projeto = fila.popleft() # Pega o 1° projeto da fila
        candidatos = preferencias[projeto]
        vagas = projetos[projeto]["vagas"]
        # Se não tem mais vaga, não olha mais
        if  len(selecionados[projeto]) >= vagas:
            continue
        # Projeto já tentou todos os candidatos
        if proximo_candidato[projeto] >= len(candidatos):
            continue

        # Passo 1: O projeto faz uma proposta
        # Seleciona o próximo aluno da lista do projeto
        aluno = candidatos[proximo_candidato[projeto]]

        # Atualiza o índice para a próxima proposta
        proximo_candidato[projeto] += 1
        historico.append((projeto, aluno, "proposta"))

        # Verifica se o aluno já possui um projeto
        proj_atual = alocado[aluno]

        # Passo 2: aluno livre
        if proj_atual is None:
            alocado[aluno] = projeto
            selecionados[projeto].append(aluno)
            historico.append((projeto, aluno, "aceito"))
        # Passo 3: O aluno já tem projeto
        else:
            preferencias_aluno = alunos[aluno]["preferencias"]
            # Quanto menor o índice, maior a preferência
            posicao_novo = preferencias_aluno.index(projeto)
            posicao_atual = preferencias_aluno.index(proj_atual)
            # Passo 3.1: Prefere o novo
            if posicao_novo < posicao_atual:
                selecionados[proj_atual].remove(aluno)
                selecionados[projeto].append(aluno)

                alocado[aluno]= projeto
                historico.append((projeto, aluno, "substituiu", proj_atual))
                # Projeto antigo volta à fila se houver candidatos
                if (proximo_candidato[proj_atual] < len(preferencias[proj_atual])):
                    fila.append(proj_atual)
            else:
                historico.append((projeto, aluno, "rejeitado_preferencia"))

        # Se ainda houver vagas, continua tentando
        if (len(selecionados[projeto]) < vagas
                and proximo_candidato[projeto] < len(candidatos)):
            fila.append(projeto)
                
    return alocado, selecionados, historico
```

```python
alocacao_p, selecionados_p, historico_p = gale_shapley_proj(alunos_limpos,projetos)

print("Alunos alocados:",
      sum(p is not None for p in alocacao_p.values()))

print("Projetos ocupados:",
      sum(len(lista) > 0 for lista in selecionados_p.values()))

print("\nEmparelhamentos:")
for projeto, lista in selecionados_p.items():
    if lista:
        print(f"{projeto}: {lista}")
```

### Verificando a estabilidade 
A verificação da estabilidade é usada para confirma se o emparelhamento feito pelo algoritmo de gale-shapley é estável, ou seja, não tem pares bloqueantes(quando um aluno e um projeto que não estão emparelhados tem interesse mútuo). Isso ocorre quando:<br>

1. O aluno prefere esse outro projeto em relação ao atual
2. O aluno tem a nota minima para passar nesse projeto
3. O projeto possui uma vaga disponível ou prefere esse aluno a algum dos alunos já selecionados.

```python
def verificar_estabilidade(alunos, projetos, alocado, selecionados):
    # Guarda se tiver algum par bloqueante
    pares_bloqueantes = []
    for aluno,dados in alunos.items():
        projeto_atual = alocado[aluno]
        for projeto in dados["preferencias"]:
            # Ignora o projeto que o aluno já ta alocado
            if projeto == projeto_atual:
                continue
            # Verifica se o aluno pode entrar no projeto
            if dados['nota'] < projetos[projeto]['nota_minima']:
                continue
            # Verifica se o aluno prefere o atual ou o outro
            if projeto_atual is not None:
                posicao_nova = dados["preferencias"].index(projeto)
                posicao_atual = dados["preferencias"].index(projeto_atual)
                if posicao_nova >= posicao_atual:
                    continue
            vagas = projetos[projeto]["vagas"]
            ocupantes = selecionados[projeto]
            # Se temos vagas sobrando, então achomos um par bloqueante
            if len(ocupantes) < vagas:
                pares_bloqueantes.append((aluno, projeto))
                continue
            # Se estiver cheio, compara com o pior aluno
            pior_aluno = min(ocupantes, key=lambda a: prioridade(a, alunos))
            # Se ele tem maior prioridade que o pior aluno, então ele 
            if prioridade(aluno, alunos) > prioridade(pior_aluno, alunos):
                pares_bloqueantes.append((aluno, projeto))

    return pares_bloqueantes
```

### Matriz de emparelhamento

```python
import pandas as pd
from IPython.display import display, HTML
# Matriz de emparelhamento para as soluções
def gerar_matriz(alunos, projetos, alocacao):
    # Lista de preferências dos projetos
    ranking_projetos = prioridade_proj(alunos, projetos)
    matriz = []

    for aluno, dados in alunos.items():
        projeto = alocacao[aluno]
        # Aluno não foi emparelhado
        if projeto is None:
            matriz.append({
                "Aluno": aluno,
                "Projeto": "Não alocado",
                "Escolha do aluno": None,
                "Classificação no projeto": None
            })
            continue

        # Posição do projeto na preferência do aluno
        escolha_aluno = dados["preferencias"].index(projeto) + 1
        # Posição do aluno na preferência do projeto
        classificacao = ranking_projetos[projeto].index(aluno) + 1
        matriz.append({
            "Aluno": aluno,
            "Projeto": projeto,
            "Escolha do aluno": escolha_aluno,
            "Classificação no projeto": classificacao
        })
    return pd.DataFrame(matriz)


# Matriz do Gale-Shapley com projetos propondo
matriz = gerar_matriz(alunos_limpos,projetos,alocacao_p)
html = matriz.to_html(index=False, na_rep="—")

display(HTML(f"""
<div style="max-height: 500px; overflow-y: auto;">
    {html}
</div>
"""))

# Matriz do Gale-Shapley com alunos propondo
matriz_a = gerar_matriz(alunos_limpos,projetos,alocacao)
html = matriz_a.to_html(index=False, na_rep="—")

display(HTML(f"""
<div style="max-height: 500px; overflow-y: auto;">
    {html}
</div>
"""))
```

### Indice de preferencia por projeto
-> Calculado a partida da quantidade de alunos que escolheram o projeto Px dividido pela quantidade total de alunos multiplicado por 100 para obter o percentual.

```python

import pandas as pd

def indice_preferencia(alunos, projetos):
    total = len(alunos)
    resultado = []
    for projeto in projetos:
        quantidade = 0
        for aluno, dados in alunos.items():
            if projeto in dados["preferencias"]:
                quantidade += 1
        percentual = (quantidade / total) * 100
        resultado.append({
            "Projeto": projeto,
            "Quantidade de preferências": quantidade,
            "Percentual de alunos (%)": round(percentual, 2)
        })

    return pd.DataFrame(resultado)

indices = indice_preferencia(alunos_limpos, projetos)

# Ordena do projeto mais procurado para o menos procurado
indices = indices.sort_values(
    by="Quantidade de preferências",
    ascending=False
).reset_index(drop=True)
display(indices)
```

### Comparação entre os dois usos de Gale-shapley

```python

def comparar_emparelhamentos(matriz_a, matriz_p):
    resultados = []
    for modalidade, df in [
        ("Alunos propondo", matriz_a),
        ("Projetos propondo", matriz_p)
    ]:
        alocados = df[df["Projeto"] != "Não alocado"]
        resultados.append({
            "Modalidade": modalidade,
            "Alunos alocados": len(alocados),
            "Projetos ocupados": alocados["Projeto"].nunique(),
            "1ª escolha": (alocados["Escolha do aluno"] == 1).sum(),
            "2ª escolha": (alocados["Escolha do aluno"] == 2).sum(),
            "3ª escolha": (alocados["Escolha do aluno"] == 3).sum(),
            "Classificação média no projeto":
                round(alocados["Classificação no projeto"].mean(), 2)
                if not alocados.empty else None
        })
    return pd.DataFrame(resultados)

comparacao = comparar_emparelhamentos(matriz_a, matriz)
display(comparacao)
```
